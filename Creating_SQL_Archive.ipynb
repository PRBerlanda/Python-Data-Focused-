```python
# Substitui valores nulos por NULL e escapa aspas
def format_value(value):
    if pd.isna(value):
        return 'NULL'
    elif isinstance(value, str):
        value = value.replace("'", "''")  # Escapa aspas simples para SQL
        return f"'{value}'"
    else:
        return str(value)

# Nome do arquivo
output_file = "alunos.sql"

# Abrir o arquivo e escrever os INSERTs
with open(output_file, 'w', encoding='utf-8') as f:
    for _, row in alunos.iterrows():
        values = (
            format_value(row['id_aluno']),
            format_value(row['nome_aluno']),
            format_value(row['data_nascimento_aluno']),
            format_value(row['genero_aluno']),
            format_value(row['endereco_aluno']),
            format_value(row['telefone_aluno']),
            format_value(row['Email'])
        )
        sql = f"INSERT INTO comentarios (id_aluno, nome_aluno, data_nascimento_aluno, genero_aluno, endereco_aluno, telefone_aluno, email_aluno) VALUES ({', '.join(values)});\n"
        f.write(sql)

print(f"Arquivo '{output_file}' gerado com sucesso.")
```

```python
# Substitui valores nulos por NULL e escapa aspas
def format_value(value):
    if pd.isna(value):
        return 'NULL'
    elif isinstance(value, str):
        value = value.replace("'", "''")  # Escapa aspas simples para SQL
        return f"'{value}'"
    else:
        return str(value)

# Nome do arquivo
output_file = "alunos.sql"

# Definir as colunas que serão inseridas na tabela 'comentarios'
insert_columns = [
    'ID_Aluno',
    'Nome_Aluno',
    'Data_Nascimento',
    'Genero',
    'Endereco',
    'Telefone_Contato',
    'Email'
]

# Abrir o arquivo e escrever os INSERTs
with open(output_file, 'w', encoding='utf-8') as f:
    for _, row in alunos.iterrows():
        # Gerar a string de labels automaticamente
        labels = ', '.join(insert_columns)

        # Gerar os valores formatados automaticamente
        values = tuple(format_value(row[col]) for col in insert_columns)

        sql = f"INSERT INTO comentarios ({labels}) VALUES ({', '.join(values)}).\n"
        f.write(sql)

print(f"Arquivo '{output_file}' gerado com sucesso.")
```
