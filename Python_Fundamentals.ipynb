## 9. Compreensões e `zip`
Sintaxe da *list comprehension*: `[expressao for item in iteravel if condicao]`. A mesma ideia vale para dicionários (`{chave: valor for ...}`) e geradores (`(expressao for ...)`). `zip` combina sequências elemento a elemento; `zip(*pares)` faz o caminho inverso.

```python
notas = [[8.0, 9.0, 10.0], [9.0, 7.0, 6.0], [3.4, 7.0, 7.0], [5.5, 6.6, 8.0], [6.0, 10.0, 9.5]]

def media(lista: list = [0]) -> float:
    calculo = sum(lista) / len(lista)
    return calculo

medias = [round(media(nota), 1) for nota in notas]
medias
```

## 12. Tratamento de exceções
`try` executa um bloco que pode falhar; `except` trata o erro (de um tipo específico ou genérico com `Exception as e`); `else` roda se não houve erro; `finally` roda sempre. `raise` lança uma exceção intencionalmente.

```python
notas = {'João': [8.0, 9.0, 10.0], 'Maria': [9.0, 7.0, 6.0], 'José': [3.4, 7.0, 8.0], 'Cláudia': [5.5, 6.6, 8.0],
         'Ana': [6.0, 10.0, 9.5], 'Joaquim': [5.5, 7.5, 9.0], 'Júlia': [6.0, 8.0, 7.0], 'Pedro': [3.0, 4.0, 6.0]}

try:
    nome = 'Mariana'  # valor de exemplo (originalmente lido com input); não está na turma -> KeyError
    resultado = notas[nome]
except KeyError:
    print('Estudante não matriculado na turma')
else:
    print(resultado)
finally:
    print('A consulta foi finalizada')
```

```python
def media(lista: list = [0]) -> float:
    calculo = sum(lista) / len(lista)
    if len(lista) > 4:
        raise ValueError('A lista não pode possuir mais de 4 notas')
    return calculo

try:
    notas = [8, 7, 9, 4, 6]
    resultado = media(notas)
except TypeError:
    print('Não foi possível calcular a média')
except ValueError as e:
    print(e)
else:
    print(resultado)
finally:
    print('A consulta foi finalizada')
```

```python
def exibir_notas_anual(lista):
    if len(lista) > 3:
        raise ValueError('A lista deve conter até 3 notas')

    for i, nota in enumerate(lista):
        print(f'{i + 1}º Trimestre = {nota}')

try:
    notas = [5.0, 2.0, 6.0, 8.0]
    exibir_notas_anual(notas)
except Exception as e:
    print(type(e), f'Erro: {e}')
```

```python
def converte_lista(lista):
    try:
        lista_float = [float(numero) for numero in lista]
    except Exception as e:
        print(type(e), f'Erro: {e}')
    else:
        return lista_float
    finally:
        print('Fim da execução')

converte_lista([1, 2, 3])
```

```python
lista1 = [4, 6, 7, 9, 10]
lista2 = [-4, 6, 8, 7, 9]

def soma_listas(lista1, lista2):
    try:
        if len(lista1) == len(lista2):
            dados = [(lista1[i], lista2[i], lista1[i] + lista2[i]) for i in range(len(lista1))]
        else:
            raise IndexError('A quantidade de elementos é diferente')
    except Exception as e:
        print(type(e), f'Erro: {e}')
    else:
        return dados

soma_listas(lista1, lista2)
```

```python
# Classificação de triângulos
lado1, lado2, lado3 = 3.0, 4.0, 4.0  # valores de exemplo (originalmente lidos com input)

if (lado1 + lado2 > lado3) and (lado2 + lado3 > lado1) and (lado1 + lado3 > lado2):
    print('Os valores podem formar um triângulo!')
    if (lado1 == lado2) and (lado2 == lado3):
        print('O triângulo é equilátero.')
    elif (lado1 != lado2) and (lado2 != lado3) and (lado1 != lado3):
        print('O triângulo é escaleno.')
    else:
        print('O triângulo é isósceles.')
else:
    print('Os valores não podem formar um triângulo!')
```

```python
# Validação de data (considerando anos bissextos)
dia, mes, ano = 29, 2, 2024  # valores de exemplo (originalmente lidos com input)

if mes == 2:
    if ano % 4 == 0 and (ano % 400 == 0 or ano % 100 != 0):
        dias_fevereiro = 29
    else:
        dias_fevereiro = 28
    if 1 <= dia <= dias_fevereiro:
        print('Data válida')
    else:
        print('Data inválida')
elif mes in [1, 3, 5, 7, 8, 10, 12]:
    if 1 <= dia <= 31:
        print('Data válida')
    else:
        print('Data inválida')
elif mes in [4, 6, 9, 11]:
    if 1 <= dia <= 30:
        print('Data válida')
    else:
        print('Data inválida')
else:
    print('Data inválida')
```

```python
# Em quantos dias a colônia A (menor, cresce mais rápido) ultrapassa a colônia B?
colonia_a = 4
colonia_b = 10
taxa_a = 0.03
taxa_b = 0.015
dias = 0

while colonia_b >= colonia_a:
    colonia_a *= 1 + taxa_a
    colonia_b *= 1 + taxa_b
    dias += 1

print(f'Irá levar {dias} dias')
print(colonia_a)
print(colonia_b)
```

```python
# Validação de notas: pede uma nova nota enquanto o valor estiver fora do intervalo
entradas = iter([5.0, 7.5, 6.0, 3.5, -1.0, 4.0])  # valores de exemplo (originalmente lidos com input)

for i in range(3):
    nota = next(entradas)
    while (nota < 0) or (nota > 6):
        print(f'Nota {i} ({nota}) é inválida. Insira uma nota válida')
        nota = next(entradas)
    print(f'Nota {i} válida: {nota}')

print('Finalizado')
```

```python
# Média de temperaturas até digitar -273 (valor de parada)
temperaturas_digitadas = iter([25.0, 27.5, 22.0, 30.5, -273])  # valores de exemplo (originalmente lidos com input)

temp = next(temperaturas_digitadas)
contador = 0
soma = 0

while temp != -273:
    soma += temp   # corrigido: o original fazia "soma + temp" e não acumulava
    contador += 1
    temp = next(temperaturas_digitadas)

media = soma / contador
print(f'A média é: {media}')
```

```python
# Pontuação de um time: vitória = 3 pontos, empate = 1
gols_marcados = [2, 1, 3, 4, 2, 0]
gols_sofridos = [0, 2, 1, 2, 4, 0]

def marca_pontos(x1, x2):
    pontos = 0
    for i in range(len(x1)):
        if x1[i] > x2[i]:      # corrigido: o original comparava as listas inteiras (x1 > x2)
            pontos += 3
        elif x1[i] == x2[i]:
            pontos += 1
    aprov = 100 * pontos / (len(x1) * 3)
    return (pontos, aprov)

pontos, aprovacao = marca_pontos(gols_marcados, gols_sofridos)
pontos, aprovacao
```

```python
# Custo de uma viagem saindo de Recife: diárias + combustível (ida e volta) + passeios
diarias = 5          # valor de exemplo (originalmente lido com input)
cidade = 'Natal'     # valor de exemplo (originalmente lido com input): 'Salvador', 'Fortaleza', 'Natal' ou 'Aracaju'
distancias = [850, 800, 300, 550]
passeio = [200, 400, 250, 300]
km_l = 14
gasolina = 5

def custo_dias(x):
    return 150 * x

def custo_combustivel(x1, x2, x3, x4):
    if x1 == 'Salvador':
        return (2 * x2[0] * x3) / x4
    elif x1 == 'Fortaleza':
        return (2 * x2[1] * x3) / x4
    elif x1 == 'Natal':
        return (2 * x2[2] * x3) / x4
    elif x1 == 'Aracaju':
        return (2 * x2[3] * x3) / x4

def custo_passeio(x1, x2, x3):
    if x1 == 'Salvador':
        return x3[0] * x2
    elif x1 == 'Fortaleza':
        return x3[1] * x2
    elif x1 == 'Natal':
        return x3[2] * x2
    elif x1 == 'Aracaju':
        return x3[3] * x2

gastos = (custo_dias(diarias),
          custo_combustivel(cidade, distancias, gasolina, km_l),
          custo_passeio(cidade, diarias, passeio))
total = sum(gastos)   # corrigido: o original exibia a tupla de gastos em vez do total
print(f'Com base nos gastos definidos, uma viagem de {diarias} dias para {cidade} saindo de Recife custaria {total:.2f} reais')
```

```python
# Tamanho, máximo, mínimo e soma de uma lista
lista = [16, 14, 63, 65, 17, 99, 70, 11, 20, 48, 79, 32, 17, 89, 12, 25, 66]

def descobertas(lista: list) -> tuple:
    tamanho = len(lista)
    maximo = max(lista)
    minimo = min(lista)
    soma = sum(lista)
    return (tamanho, maximo, minimo, soma)

tamanho, maximo, minimo, soma = descobertas(lista)
print(f'A lista possui {tamanho} itens, valor mínimo {minimo}, valor máximo {maximo} e somam-se {soma}')
```

```python
# Múltiplos de 3 de uma lista
numeros = [97, 80, 94, 88, 80, 1, 16, 53, 62, 32, 24, 99]

def multiplos(lista: list) -> list:
    '''Retorna os elementos da lista que são múltiplos de 3 (resto da divisão por 3 igual a 0).'''
    mult_3 = []
    for numero in lista:
        if numero % 3 == 0:
            mult_3.append(numero)
    return mult_3

multiplos(numeros)
```

### 13.4 Exceções

```python
# Corretor de provas: lança ValueError se houver alternativa inválida
gabarito = ['D', 'A', 'B', 'C', 'A']

def corretor(testes: list):
    pontuacoes = []   # lista que receberá as pontuações caso a exceção não seja lançada
    try:
        for teste in testes:
            nota = 0  # acumula a nota de cada estudante
            for i in range(len(teste)):
                if teste[i] not in ['A', 'B', 'C', 'D']:   # alternativa válida?
                    raise ValueError(f'A alternativa {teste[i]} não é uma opção de alternativa válida')
                elif teste[i] == gabarito[i]:              # resposta igual ao gabarito soma 1 ponto
                    nota += 1
            pontuacoes.append(nota)
    except Exception as e:
        print(e)
    else:
        return pontuacoes

# Exemplo que não lança exceção
testes_sem_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'C', 'A'], ['D', 'B', 'A', 'C', 'A']]
print(corretor(testes_sem_ex))

# Exemplo que lança exceção (alternativa 'E')
testes_com_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'E', 'C', 'A']]
corretor(testes_com_ex)
```

```python
# Verifica se as palavras de um texto já tratado ainda contêm pontuação
sinais_pontuacao = [',', '.', '!', '?']

def verifica_pontuacao(texto: list):
    for palavra in texto:
        if any(sinal in palavra for sinal in sinais_pontuacao):
            raise ValueError(f'A palavra "{palavra}" contém pontuação')
    return 'Texto tratado!'

lista_tratada = ['Python', 'é', 'uma', 'linguagem', 'de', 'programação', 'poderosa', 'versátil',
                 'e', 'fácil', 'de', 'aprender', 'utilizada', 'em', 'diversos', 'campos', 'desde',
                 'análise', 'de', 'dados', 'até', 'inteligência', 'artificial']

lista_nao_tratada = ['Python', 'é', 'uma', 'linguagem', 'de', 'programação', 'poderosa,', 'versátil',
                     'e', 'fácil,', 'de', 'aprender', 'utilizada', 'em', 'diversos', 'campos,', 'desde',
                     'análise', 'de', 'dados', 'até', 'inteligência', 'artificial!']

for texto in [lista_tratada, lista_nao_tratada]:
    try:
        avalia_texto = verifica_pontuacao(texto)
    except Exception as e:
        print(e)
    else:
        print(avalia_texto)
```

```python
# Razão pressão / temperatura com tratamento de ValueError e ZeroDivisionError
def divide_colunas(pressao: list, temperatura: list) -> list:
    try:
        if len(pressao) != len(temperatura):
            raise ValueError('A pressão e a temperatura contêm listas de tamanhos diferentes')
        resultado = [round(a / b, 2) for a, b in zip(pressao, temperatura)]
    except ValueError as e:
        print(e)
    except ZeroDivisionError as e:
        print(f'{e}: o valor não pode ser 0')
    else:
        return resultado

# Exemplo que não lança exceção
print(divide_colunas([100, 120, 140, 160, 180], [20, 25, 30, 35, 40]))

# Exemplo que lança ZeroDivisionError
divide_colunas([60, 120, 140, 160, 180], [0, 25, 30, 35, 40])

# Exemplo que lança ValueError
divide_colunas([100, 120, 140, 160], [20, 25, 30, 35, 40])
```
