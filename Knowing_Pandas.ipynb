```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv'
df = pd.read_csv(url, sep = ';')
df.head(10)
df.tail(10)
```

```python
# Conhecendo as informações do dataFrame

type(df) # Mostra o tipo do dado (nesse caso, um dataFrame)

df.shape # Formato do dataFrame, quantidade de linhas e colunas

df.columns # Nome das colunas

df.info() # Contém mais informações sobre o dataFrame

df['Tipo'] # Estrutura do tipo 'SERIES', onde há somente uma coluna com índices

df[['Quartos', 'Valor']] # São utilizados dois colchetes pois é preciso transformar em um dataFrame [] e depois inserir as colunas no próximo []

df.describe() # Informações estatísticas sobre as colunas numéricas
```

```python
df_preco_tipo = df.groupby('Tipo')[['Valor']].mean().sort_values('Valor')
 # Agrupa a coluna 'Tipo', fazendo a média apenas da coluna 'Valor', ordenando por valor crescente.
 # Note que foi inserido colchetes duplos [[]], criando um dataFrame para poder agrupar os dados.

df_preco_tipo.plot(kind = 'barh', figsize=(16, 10), color='purple')
```

```python
df.Tipo.unique()
imoveis_comerciais= ['Conjunto Comercial/Sala',
                     'Prédio Inteiro',
                     'Loja/Salão',
                     'Galpão/Depósito/Armazém',
                     'Casa Comercial',
                     'Terreno Padrão',
                     'Loja Shopping/ Ct Comercial',
                     'Box/Garagem',
                     'Chácara',
                     'Loteamento/Condomínio',
                     'Sítio',
                     'Pousada/Chalé',
                     'Hotel',
                     'Indústria']
df = df.query('@imoveis_comerciais not in Tipo') # Inserir um @ quando quer referenciar uma variável, assim como no SQL.
df.Tipo.unique()
```

```python
df_preco_tipo = df.groupby('Tipo')[['Valor']].mean().sort_values('Valor').plot(kind = 'barh', figsize=(16, 10), color='purple')
```

```python
# Contando vezes que o valor aparece, fazendo o sort e incluindo em um dataframe
df_percentual_tipo = df.Tipo.value_counts(normalize=True).to_frame().sort_values('Tipo')
# Alterando o nome da coluna de porcentagem
df_percentual_tipo.rename(columns={'proportion':'Percentuais'}, inplace=True)
# Plotando a figura
df_percentual_tipo.plot(kind='bar', figsize=(14,10), color = 'green', edgecolor = 'white', xlabel = 'Tipo', ylabel = 'Percentual')
```

```python
df = df.query('Tipo == "Apartamento"')
df.Tipo.unique()
```

```python
df.Quartos.mean()
len(df['Bairro'].unique())
df.Valor.mean()
media_bairros = df.groupby('Bairro')[['Valor']].mean(numeric_only=True).sort_values('Valor', ascending=False).head()
bairros_maiores_que_media = media_bairros.query('Valor > 4744.614647887324')
bairros_maiores_que_media.plot(kind='barh', figsize=(14, 10), color='red', xlabel='Valor', ylabel='Bairro')
```

```python
df.isnull().sum() # soma quantos valores são nulos
df = df.fillna(0) # preenche os nulos com o valor 0
df.fillna(method='ffill', inplace=True) # preenche o valor nulo com o valor acima dele
df.fillna(method='bfill', inplace=True) # preenche o valor nulo com o valor abaixo dele
df.interpolate(method='linear', axis=0, inplace=True) # preenche valores nulos com o cálculo dos valores vizinhos
```

```python
registros_a_remover = df.query('Valor == 0 | Condominio == 0').index # Verifica apartamentos sem valor ou sem condomínio
df.drop(registros_a_remover, axis=0, inplace=True) # Faz a remoção das linhas
df.drop('Tipo', axis=1, inplace=True) # Remove a coluna 'Tipo' pois só há um tipo de categoria (apartamento)
df
```

```python
selecao1 = df['Quartos'] == 1
selecao2 = df['Valor'] < 1200
selecao_final = (selecao1) & (selecao2)
df_selecionada = df[selecao_final]
df_selecionada

df_selecionada_2 = (df['Quartos'] >= 2) & (df['Valor'] < 1200) & (df['Area'] > 70)
selecao_final_2 = df[df_selecionada_2]
selecao_final_2.head()

df_selecionada_3 = (df['Area'] > 80) | (df['Valor'] < 4000)
selecao_final_3 = df[df_selecionada_3]
selecao_final_3
```

```python
df.to_csv('dados_apartamentos.csv', index=False, sep=';')
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv'
df = pd.read_csv(url, sep = ';')
```

```python
df['Valor_por_mes'] = df['Valor'] + df['Condominio'] # Soma o valor de condomínio mais o valor do aluguel
df['Valor_por_ano'] = df['Valor_por_mes'] * 12 + df['IPTU']
df.head()
```

```python
df['Descricao'] = df['Tipo'] + ' em ' + df['Bairro'] + ' com ' + df['Quartos'].astype(str) + ' quarto(s) ' + \
' e ' + df['Vagas'].astype(str) + ' vaga(s) de garagem.'
df.Descricao.head()
```

```python
df['Possui_suite'] = df['Suites'].apply(lambda x: 'Sim' if x > 0 else 'Não')
df.Possui_suite.head()
```

```python
df.to_csv('dados_completos_dev.csv', index=False, sep=';')
```
