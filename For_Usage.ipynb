```python
gols_marcadosx1 = [2, 1, 3, 4, 2, 0]
gols_sofridosx2 = [0, 2, 1, 2, 4, 0]

def marca_pontos(x1, x2):
  pontos = 0
  for i in range(len(x1)):
    if x1 > x2:
      pontos += 3
    elif x1 == x2:
      pontos += 1
    else:
      pontos += 0
  aprov = 100 * pontos / (len(x1) * 3)
  return (pontos, aprov)

pontos, aprovacao = marca_pontos(gols_marcadosx1, gols_sofridosx2)
pontos, aprovacao
```

```python
diarias = int(input("Diárias: "))
cidades = input("Qual a cidade? ['Salvador', 'Fortaleza', 'Natal', 'Aracaju']: ")
distancias = [850, 800, 300, 550]
passeio = [200, 400, 250, 300]
km_l = 14
gasolina = 5

def custo_dias(x):
  return 150 * x

def custo_combustivel(x1, x2, x3, x4):
  if x1 == 'Salvador':
    return (2 * x2[0] * x3) / x4
  elif x1 == 'Fortaleza':
    return (2 * x2[1] * x3) / x4
  elif x1 == 'Natal':
    return (2 * x2[2] * x3) / x4
  elif x1 == 'Aracaju':
    return (2 * x2[3] * x3) / x4

def custo_passeio(x1, x2, x3):
    if x1=="Salvador":
        return x3[0] * x2
    elif x1=="Fortaleza":
        return x3[1] * x2
    elif x1=="Natal":
        return x3[2] * x2
    elif x1=="Aracaju":
        return x3[3] * x2

gastos = custo_dias(diarias), custo_combustivel(cidades, distancias, gasolina, km_l), custo_passeio(cidades, diarias, passeio)
gastos
print(f"Com base nos gastos definidos, uma viagem de {diarias} dias para {cidades} saindo de Recife custaria {gastos} reais")
```
