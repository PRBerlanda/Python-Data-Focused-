```python
import pandas as pd

hospedagens = pd.read_json('/content/dados_hospedagem.json')
moveis = pd.read_json('/content/moveis_disponiveis.json')
```

```python
hospedagens = pd.json_normalize(hospedagens['info_moveis'])
```

```python
colunas = list(hospedagens.columns)
colunas
```

```python
hospedagens = hospedagens.explode(colunas[3:])
hospedagens.reset_index(inplace=True, drop=True)
```

```python
hospedagens.info()
```

```python
import numpy as np
```

Manipulação de números

```python
hospedagens['max_hospedes'] = hospedagens['max_hospedes'].astype(np.int64)
hospedagens['avaliacao_geral'] = hospedagens['avaliacao_geral'].astype(np.float64)
```

```python
col_numericas = ['quantidade_banheiros', 'quantidade_quartos', 'quantidade_camas']
hospedagens[col_numericas] = hospedagens[col_numericas].astype(np.int64)
```

```python
hospedagens['preco'] = hospedagens['preco'].apply(lambda x: x.replace('$','').replace(',','').strip())
```

```python
hospedagens[['taxa_deposito', 'taxa_limpeza']] = hospedagens[['taxa_deposito', 'taxa_limpeza']].applymap(lambda x: x.replace('$','').replace(',','').strip())
```

```python
hospedagens[['taxa_deposito', 'taxa_limpeza']] = hospedagens[['taxa_deposito', 'taxa_limpeza']].astype(np.float64)
```

Manipulação de textos

```python
hospedagens['descricao_local'] = hospedagens['descricao_local'].str.lower()
```

```python
hospedagens['descricao_local'] = hospedagens['descricao_local'].str.replace('[^a-zA-Z0-9\-\']',' ',regex=True)
hospedagens['descricao_local'] = hospedagens['descricao_local'].str.replace('(?<!\w)-(?!\w)',' ',regex=True)
```

```python
hospedagens['descricao_local'] = hospedagens['descricao_local'].str.split()
```

```python
hospedagens['comodidades'] = hospedagens['comodidades'].str.replace('\{|}|\"','', regex=True)
```

```python
hospedagens['comodidades'] = hospedagens['comodidades'].str.split(',')
hospedagens.head()
```

```python
hospedagens['descricao_vizinhanca'] = hospedagens['descricao_vizinhanca'].str.lower().replace(r'[^a-zA-Z0-9\-\'\--]', ' ', regex=True).replace(r'(?<!\w)-(?!\w)', ' ', regex=True).str.split()
```

```python
loc_imoveis = pd.read_json('/content/dados_locacao_imoveis.json')
```

```python
clientes = pd.read_csv('/content/banco_clientes.csv', sep=',')
```

```python
clientes['nome'] = clientes['nome'].str.upper()
```

```python
clientes['telefone'] = clientes['telefone'].replace(r'\(|\)', '',regex=True)
```

```python
clientes['cep'] = clientes['endereco'].str.extract(r'(\d{5}-\d{3})')
```

```python
clientes['endereco'] = clientes['endereco'].str.replace(r'(\d{5}-\d{3})', '', regex=True).str.strip()
```

```python
clientes[["rua", "numero"]] = clientes['endereco'].str.replace(',','').str.split('\s(?=\d+$)', regex=True, expand=True)
```
