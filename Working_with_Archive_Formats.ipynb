**Importando BANCO DE DADOS**

```python
import sqlalchemy

# 'create_engine' cria o motor que vai fazer a conexão com o banco de dados
# 'MetaData' representa os metadados do DB, sendo informações sobre tabelas, colunas e outros objetos
# 'inspect' inspeciona a estrutura do DB
from sqlalchemy import create_engine, MetaData, Table, inspect

# Cria um banco de dados SQLite local ('memory')
motor = create_engine('sqlite:///:memory:')

url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/refs/heads/main/clientes_banco.csv'

df = pd.read_csv(url)
df.to_sql('clientes', motor, index=False)
df

# Inspeciona o motor que faz a consulta rodar e printa a tabela que está lá dentro
inspector = inspect(motor)
print(inspector.get_table_names())

# Fazendo uma consulta SQL

query = 'SELECT * FROM clientes WHERE Categoria_de_renda = "Empregado"'
empregados = pd.read_sql(query, motor)

empregados.to_sql('empregados', con=motor, index=False)

# read_sql lê uma consulta, enquanto read_sql_table lê uma tabela completa
pd.read_sql_table('empregados', motor)

pd.read_sql_table('empregados', motor, columns=['ID_Cliente', 'Grau_escolaridade', 'Rendimento_anual'])
```
