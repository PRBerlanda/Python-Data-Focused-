# Parte 1 — Matplotlib, Seaborn e Plotly

## 1. Dados de imigração
Número anual de imigrantes para o Canadá por país de origem (1980 a 2013). O país vira índice e as colunas de ano são usadas para montar as séries.

```python
df = pd.read_csv('imigrantes_canada.csv', delimiter=',', encoding='utf8')
df.set_index('País', inplace=True)
df.head()
```

```python
# Colunas de ano (de 1980 a 2013, inclusive)
anos = list(map(str, range(1980, 2014)))

brasil = df.loc['Brasil', anos]
dados_brasil = pd.DataFrame({'ano': brasil.index.tolist(),
                             'imigrantes': brasil.values.tolist()})
dados_brasil
```

### 1.1 Alternativa com `melt`
O `melt` transforma as colunas de ano em linhas (formato longo), o que facilita filtrar vários países de uma vez.

```python
imig_longo = pd.melt(df.reset_index(), id_vars='País', value_vars=anos,
                     var_name='Ano', value_name='Imigrantes')
brasil_argentina = imig_longo[imig_longo['País'].isin(['Brasil', 'Argentina'])].set_index('País')
brasil_argentina
```

## 2. Matplotlib
Gráficos de linha, subplots, boxplot, estilos e barras com destaque, primeiro pela interface `plt` e depois pela orientada a objetos (`fig, ax`).

### 2.1 Gráfico de linha com `plt`

```python
plt.figure(figsize=(16, 10))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.title('Imigração do Brasil para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Imigrantes')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.yticks([500, 1000, 1500, 2000, 2500, 3000])
plt.show()
```

### 2.2 Duas séries com legenda

```python
plt.figure(figsize=(16, 10))
plt.plot(brasil_argentina.loc['Brasil']['Ano'], brasil_argentina.loc['Brasil']['Imigrantes'], label='Brasil')
plt.plot(brasil_argentina.loc['Argentina']['Ano'], brasil_argentina.loc['Argentina']['Imigrantes'], label='Argentina')
plt.title('Imigração do Brasil e Argentina para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Imigrantes')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.legend()
plt.show()
```

```python
# Relatório de vendas, com as datas de pedido e envio convertidas para datetime
vendas = pd.read_csv('https://raw.githubusercontent.com/alura-cursos/dataviz-graficos/master/dados/relatorio_vendas.csv')
vendas['data_pedido'] = pd.to_datetime(vendas['data_pedido'])
vendas['data_envio'] = pd.to_datetime(vendas['data_envio'])

vendas.info()
```

### 5.1 Gráfico de colunas — Pergunta 1: total de vendas por ano

```python
df_vendas_ano = vendas[['data_pedido', 'vendas']].copy()
df_vendas_ano['ano'] = df_vendas_ano['data_pedido'].dt.year
df_vendas_ano = df_vendas_ano.drop(columns='data_pedido').groupby('ano').aggregate('sum')
df_vendas_ano
```

```python
def grafico_vendas(cores: list = [AZUL2]):
    # Tema e área do gráfico
    sns.set_theme(style='white')
    fig, ax = plt.subplots(figsize=(10, 4))

    # Gráfico de colunas
    ax = sns.barplot(data=df_vendas_ano, x=df_vendas_ano.index, y='vendas', hue='vendas', legend=False, palette=cores)

    # Personalização
    ax.set_title('Vendas de 2016 a 2019', loc='left', fontsize=17, color=CINZA1)
    ax.set_xlabel('Ano')
    ax.set_ylabel('Vendas')
    ax.yaxis.set_tick_params(labelsize=10, labelcolor=CINZA2)
    sns.despine(left=True, bottom=True)

    # Valores escritos sobre cada coluna (no lugar do eixo y)
    ax.set_yticklabels([])
    for i, valor in enumerate(df_vendas_ano['vendas']):
        qtd = f'R$ {valor:,.0f}'.replace(',', '.')
        offset = 1e5
        ax.text(i, valor + offset, qtd, color=CINZA2, fontsize=12, ha='center', va='center')

    return ax

ax = grafico_vendas()
```

### 5.2 Adicionando texto na visualização
Destaque por cor para o ano de maior venda e uma conclusão escrita ao lado do gráfico.

```python
cores = [AZUL2 if v == df_vendas_ano['vendas'].max() else AZUL5 for v in df_vendas_ano['vendas']]

ax = grafico_vendas(cores)
ax.text(3.5, 1.5e6,
        'Em $\\bf{2019}$, as vendas\n'
        'nas lojas subiram\n'
        'aproximadamente $\\bf{22,3}$%\n'
        'em relação ao ano de 2018.',
        fontsize=14, linespacing=1.45, color=AZUL2)
plt.show()
```

**Conclusão:** o gráfico de colunas é ideal para comparar poucas categorias ou poucos períodos. O destaque por cor guia o olhar para o último período, e o texto explica a evolução em relação ao ano anterior. A mesma técnica serve para o lucro anual ou o faturamento por estado/região.

### 5.3 Gráfico de barras — Pergunta 2: top 7 produtos em lucro

```python
top_7 = (vendas[['tipo_produto', 'lucro']]
         .groupby('tipo_produto').aggregate('sum')
         .sort_values('lucro', ascending=False).head(7))
top_7
```

```python
def grafico_top_7():
    sns.set_theme(style='white')
    fig, ax = plt.subplots(figsize=(10, 4))
    fig.subplots_adjust(right=0.7)

    # Cores por departamento: Automotivo (azul), Jardinagem e paisagismo (verde), outro (cinza)
    cores = [AZUL2, AZUL2, AZUL2, VERDE1, VERDE1, CINZA3, VERDE1]
    ax = sns.barplot(data=top_7, x='lucro', y=top_7.index, hue=top_7.index, legend=False, palette=cores)

    ax.set_title('Top 7 produtos com maior lucro no catálogo (2016-2019)\n', fontsize=18, color=CINZA1, loc='left')
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_xticklabels([])
    ax.yaxis.set_tick_params(labelsize=10, labelcolor=CINZA2)
    sns.despine(left=True, bottom=True)

    # Valores dentro das barras
    for i, valor in enumerate(top_7['lucro']):
        qtd = f'R$ {valor:,.0f}'.replace(',', '.')
        offset = 1e3
        ax.text(valor - offset, i, qtd, color=CINZA5, fontsize=10, fontweight='bold', ha='right', va='center')

    return ax

ax = grafico_top_7()
```

### 5.4 Texto com destaques coloridos
Não existe função nativa para escrever um texto com várias cores. A função abaixo (adaptada de [storytelling-with-data](https://github.com/empathy87/storytelling-with-data)) separa os trechos de cada linha por `||` e recebe uma lista de cores por linha.

```python
def texto_colorido(x, y, texto, cores, esp=20, ax=None, **kw):
    cores = list(reversed(cores))
    t = ax.transData
    canvas = ax.figure.canvas

    for i, linha in enumerate(reversed(texto.split('\n'))):
        frases = linha.split('||')
        for s, cor in zip(frases, cores[i]):
            texto = ax.text(x, y, s, color=cor, transform=t, **kw)
            texto.draw(canvas.get_renderer())
            ex = texto.get_window_extent()
            t = transforms.offset_copy(texto._transform, x=ex.width, units='dots')

        t = transforms.offset_copy(ax.transData, x=0, y=(i + 1) * esp, units='dots')
```

```python
ax = grafico_top_7()

texto_colorido(
    9.2e4, 3.25,                                                   # coordenadas
    'Os dados indicam que os 3 produtos que geram\n'               # texto
    '$\\bf{maior\\ lucro}$|| são do departamento ||$\\bf{Automotivo}$.\n'
    '\n'
    'Podemos notar também que o departamento de\n'
    '$\\bf{Jardinagem\\ e\\ paisagismo}$|| possui 3 produtos com\n'
    'uma boa margem de lucro, sendo que um deles está\n'
    'abaixo de ||$\\bf{50\\ mil\\ reais}$|| no período mencionado.',
    [[CINZA2],                     # linha 1 (cores)
     [CINZA1, CINZA2, AZUL2],      # linha 2
     [CINZA2],                     # linha 3
     [CINZA2],                     # linha 4
     [VERDE1, CINZA2],             # linha 5
     [CINZA2],                     # linha 6
     [CINZA2, CINZA1, CINZA2]],    # linha 7
    esp=22, ax=ax, fontsize=10)
plt.show()
```

**Conclusão:** o gráfico de barras é ideal para comparar muitos itens (acima de 5), principalmente com rótulos em palavras ou frases. As cores separam categorias (departamentos) e o texto repete esse destaque.

## 6. Agrupando dados: colunas e barras empilhadas
### 6.1 Colunas empilhadas — Pergunta 3: vendas por ano e por região

## 8. Distribuindo dados: histograma, densidade e boxplot
**Situação-problema 2:** uma fábrica de produtos de higiene e limpeza mede regularmente amostras dos seus produtos para garantir os critérios de qualidade. O objetivo é descrever as amostras (média, mediana, quartis) com gráficos. Bases:

2. Como se distribuem de acordo com a mediana e os quartis? Há dados discrepantes?

### 8.1 Histograma — Pergunta 1
O número de intervalos (`bins`) ou a sua largura (`binwidth`) muda um pouco a forma do histograma, mas não os valores centrais.

```python
sns.set_theme(style='white')
fig, ax = plt.subplots(figsize=(12, 4))

# Histograma sem curva de densidade
ax = sns.histplot(data=vol_amaciante, x='Volume', bins=20, color=AZUL2, kde=False)  # quantidade de bins
# ax = sns.histplot(data=vol_amaciante, x='Volume', binwidth=0.1, color=AZUL2)      # alternativa: largura dos bins

plt.suptitle('Distribuição dos volumes do amaciante', size=18, color=CINZA1, ha='right', x=0.535, y=1.05)
plt.title('para uma amostra de 1000 unidades', fontsize=14, color=CINZA2, pad=15, loc='left')
ax.set_xlabel('Volume (ml)', fontsize=14)
ax.set_ylabel('Quantidade', fontsize=14)
ax.yaxis.set_tick_params(labelsize=12, labelcolor=CINZA2)
ax.xaxis.set_tick_params(labelsize=12, labelcolor=CINZA2)
sns.despine()
plt.show()
```

### 8.2 Gráfico de densidade
A curva de densidade é um "histograma de linha", indicada para muitas observações.

```python
def grafico_densidade():
    sns.set_theme(style='white')
    fig, ax = plt.subplots(figsize=(15, 4))

    # Curva de densidade
    ax = sns.kdeplot(data=vol_amaciante, x='Volume', color=AZUL2, fill=True)

    plt.suptitle('Distribuição dos volumes do amaciante (curva de densidade)', size=18, color=CINZA1, ha='right', x=0.635, y=1.05)
    plt.title('para uma amostra de 1000 unidades', fontsize=14, color=CINZA2, pad=15, loc='left')
    ax.set_xlabel('Volume', fontsize=14)
    ax.set_ylabel('Densidade', fontsize=14)
    ax.xaxis.set_tick_params(labelsize=12, labelcolor=CINZA2)
    ax.yaxis.set_tick_params(labelsize=12, labelcolor=CINZA2)
    ax.set_xlim(vol_amaciante['Volume'].min(), vol_amaciante['Volume'].max())
    sns.despine(left=True, top=True)

    return ax

ax = grafico_densidade()
```

### 8.3 Adicionando linhas ao gráfico
Linhas verticais marcam média, mediana e moda, com os valores anotados.

```python
ax = grafico_densidade()
ax.set_ylabel('')
ax.set_yticklabels([])

# Média, mediana e moda
desc = ['Média', 'Mediana', 'Moda']
estat = [round(vol_amaciante['Volume'].mean(), 2), vol_amaciante['Volume'].median(), vol_amaciante['Volume'].mode()[0]]
cores = [LARANJA1, VERDE1, CINZA3]

# Uma linha vertical para cada valor de x
ax.vlines(x=estat, ymin=0, ymax=ax.get_ylim()[1], colors=cores, linestyles='--', lw=2)

for i in range(len(estat)):
    ax.text(0.75, (0.6 - i / 10), f'{desc[i]} = {estat[i]} ml', fontsize=14, color=cores[i], transform=ax.transAxes)

plt.show()
```

**Conclusão:** a distribuição é aproximadamente normal, com leve assimetria à esquerda (mediana > média). Média, mediana e moda são as medidas de tendência central da estatística descritiva.

### 8.4 Boxplot — Pergunta 2
O boxplot resume os dados em 5 medidas: mínimo, 1º quartil (25%), mediana (50%), 3º quartil (75%) e máximo; os pontos além dos "bigodes" são os *outliers* (valores discrepantes).

```python
sns.set_theme(style='white')
fig, ax = plt.subplots(figsize=(6, 8))

ax = sns.boxplot(data=vol_amaciante, y='Volume', color=AZUL2, width=0.3)

plt.suptitle('Boxplot do volume do amaciante', size=18, color=CINZA1, ha='right', x=0.81, y=0.97)
plt.title('para uma amostra de 1000 unidades', fontsize=14, color=CINZA2, pad=15, loc='left')
ax.set_xlabel('')
ax.set_ylabel('Volume (ml)', fontsize=14)
ax.yaxis.set_tick_params(labelsize=12, labelcolor=CINZA2)
sns.despine(bottom=True)

# Elipses em volta dos outliers
y_min, y_max = ax.get_ybound()
ax.add_patch(Ellipse((0, y_max - 0.27), width=0.1, height=0.46, color=VERMELHO1, fill=False))
ax.add_patch(Ellipse((0, y_min + 0.4), width=0.1, height=0.6, color=VERMELHO1, fill=False))

# Explicação dos elementos do boxplot
for y in [y_max - 0.27, y_min + 0.4]:
    ax.annotate('Outliers', xy=(0 + 0.05, y), xycoords='data',
                bbox=dict(boxstyle='round', fc=CINZA5, ec=VERMELHO1),
                xytext=(50, 0), textcoords='offset points',
                arrowprops=dict(arrowstyle='->', color=VERMELHO1))

for rotulo, valor in [('1º Quartil / 25%', vol_amaciante['Volume'].quantile(.25)),
                      ('2º Quartil / Mediana', vol_amaciante['Volume'].median()),
                      ('3º Quartil / 75%', vol_amaciante['Volume'].quantile(.75))]:
    ax.annotate(rotulo, xy=(0.4, valor), xycoords='data',
                bbox=dict(boxstyle='round', fc=CINZA5, ec=CINZA3),
                xytext=(50, 0), textcoords='offset points',
                arrowprops=dict(arrowstyle='->', color=CINZA3))

plt.show()
```

### 9.2 Gráfico de dispersão — Pergunta 2
Comprimento × altura da amostra A, com as linhas dos limites de aceitação.

```python
df_a = df_medidas.query("amostra == 'A'")[['comprimento', 'altura']]
df_a.head()
```

```python
def grafico_dispersao(map_cores: list = [AZUL2], pct: float = 0.01):
    sns.set_theme(style='white')
    fig, ax = plt.subplots(figsize=(10, 5))

    # Dispersão: uma cor para todos os pontos ou uma cor por ponto
    ax.scatter(df_a['comprimento'], df_a['altura'], c=map_cores)

    plt.suptitle('Distribuição do comprimento x altura da caixa de sabão em pó', size=18, color=CINZA1, ha='right', x=0.91, y=1.03)
    plt.title('Para a amostra A com 200 observações', fontsize=14, color=CINZA2, pad=15, loc='left')
    ax.set_xlabel('Comprimento (cm)', fontsize=14)
    ax.set_ylabel('Altura (cm)', fontsize=14)
    ax.xaxis.set_tick_params(labelsize=12, labelcolor=CINZA2)
    ax.yaxis.set_tick_params(labelsize=12, labelcolor=CINZA2)
    sns.despine()

    # Linha horizontal: limite mínimo de altura
    ax.text(19.92, 14.855, 'Limite mínimo da altura', fontsize=12, color=CINZA2, va='bottom')
    ax.axhline(y=15 * (1 - pct), color=CINZA4, linestyle='--')

    # Linha vertical: limite máximo de comprimento
    ax.text(20.205, 15.05, 'Limite máximo\nde comprimento', fontsize=12, color=CINZA2, ha='left')
    ax.axvline(x=20 * (1 + pct), color=CINZA4, linestyle='--')

    return ax

ax = grafico_dispersao()
```

### 9.3 Mapeando dados por cores
Cada ponto recebe uma cor conforme a regra de rejeição: rejeitado por comprimento **e** altura (vermelho), por um dos dois (laranja) ou aceito (azul).

```python
cores = [VERMELHO1, LARANJA1, AZUL2]

pct = 0.01
rejeita_comp = [c > 20 * (1 + pct) or c < 20 * (1 - pct) for c in df_a['comprimento']]
rejeita_alt = [a > 15 * (1 + pct) or a < 15 * (1 - pct) for a in df_a['altura']]
map_cores = [cores[0] if (c and a) else cores[1] if (c or a) else cores[2]
             for c, a in zip(rejeita_comp, rejeita_alt)]

ax = grafico_dispersao(map_cores, pct)

# Áreas de rejeição
ax.add_patch(Ellipse((20.165, 14.80), width=0.065, height=0.08, color=CINZA1, fill=False))
ax.add_patch(Ellipse((20.24, 14.715), width=0.04, height=0.08, color=CINZA1, fill=False))

ax.annotate('Rejeição pela faixa\nde altura', xy=(20.185, 14.82), xycoords='data',
            bbox=dict(boxstyle='round', fc=CINZA5, ec=CINZA3),
            xytext=(50, -20), textcoords='offset points',
            arrowprops=dict(arrowstyle='->', color=CINZA3))
ax.annotate('Rejeição pela faixa\nde comprimento e altura', xy=(20.22, 14.72), xycoords='data',
            bbox=dict(boxstyle='round', fc=CINZA5, ec=CINZA3),
            xytext=(-180, -20), textcoords='offset points',
            arrowprops=dict(arrowstyle='->', color=CINZA3))

plt.show()
```

**Conclusão:** o gráfico de dispersão mostra como as dimensões se distribuem entre duas variáveis. Os limites desenhados delimitam a faixa aceita, e as cores agrupam os itens aceitos e os rejeitados por cada regra.
